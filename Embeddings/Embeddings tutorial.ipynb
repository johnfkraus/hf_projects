```python
import seaborn as sns
import matplotlib.pyplot as plt

# Example DataFrame
df = pd.DataFrame({'gender': [1,9,1,9], 'age': [7,7,2,2], 'index_label': ['man', 'woman', 'boy', 'girl']})

print(df)
# Create the scatter plot
ax = sns.scatterplot(data=df, x='gender', y='age')

# Annotate points using the DataFrame index
for idx, row in df.iterrows():
    ax.text(row['gender'], row['age'], row['index_label'], ha='center', va='bottom')

plt.show()   
```
